```python
import duckdb

con = duckdb.connect('/home/DAVIDSON/pagrimes/rlim-project/data/transit_warehouse.duckdb')
con.execute("LOAD spatial;")

# Verify spatial parcels join with standardized sales
df = con.execute("""
    SELECT 
        p.parcel_id,
        p.transit_buffer_tier,
        p.nearest_station_name,
        p.dist_to_station_meters,
        s.real_price_2026,
        s.ppsf_real_2026
    FROM silver.parcels_spatial p
    INNER JOIN silver.sales_standardized s 
        ON p.parcel_id = s.parcel_id
    LIMIT 10;
""").fetchdf()

display(df)
con.close()
```
